```python
import pandas as pd
import numpy as np

URL = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv"
# Output dari pd.read_csv adalah DataFrame
df = pd.read_csv(URL)
print("Dataset berhasil dimuat.")
print("Shape dataset:", df.shape)
```

```python
# Berapa jumlah baris dan kolom dataset Tips?
# Dapat dijawab menggunakan DataFrame.shape, output-nya (jumlah_baris, jumlah_kolom)
df_shape = df.shape
jumlah_baris = df_shape[0]
jumlah_kolom = df_shape[1]
print('Jumlah baris:', jumlah_baris)
print('Jumlah kolom:', jumlah_kolom)
```

```python
# Kolom mana yang memiliki nilai kosong (missing)?
# Dapat dijawab menggunakan DataFrame.info()
print(df.info())
```

```python
# Berapa persen pelanggan yang merokok (smoker = 'Yes')?
# Tahapannya:
# 1. Filter baris dengan kondisi smoker == 'Yes'
# 2. Hitung rasio terhadap total baris dan kalikan 100
smokers = df[df['smoker'] == 'Yes'].shape[0]
total = df.shape[0]
persen_smoker = (smokers / total) * 100
print(f"Jumlah pelanggan merokok: {smokers} dari {total}")
print(f"Persentase pelanggan merokok: {persen_smoker:.2f}%")
```

```python
# Rata-rata tip yang diberikan oleh pelanggan wanita (sex = 'Female') pada waktu makan malam (time = 'Dinner')?
# Tahapannya:
# 1. Buat DataFrame terfilter dengan kondisi sex == 'Female' dan time == 'Dinner'
# 2. Ambil kolom 'tip' dan hitung rata-rata menggunakan fungsi .mean()
df_filtered = df[(df['sex'] == 'Female') & (df['time'] == 'Dinner')]
avg_tip_female_dinner = df_filtered['tip'].mean()
print(f"Jumlah transaksi wanita saat dinner: {df_filtered.shape[0]}")
print(f"Rata-rata tip: ${avg_tip_female_dinner:.2f}")
```

```python
# Hari apa yang rata-rata total tagihannya (total_bill) paling tinggi?
# Tahapannya:
# 1. Lakukan grouping berdasarkan kolom 'day' menggunakan groupby()
# 2. Ambil kolom 'total_bill' dan panggil fungsi mean()
avg_bill_per_day = df.groupby('day')['total_bill'].mean().sort_values(ascending=False)
print("Rata-rata total bill per hari:")
print(avg_bill_per_day)
```
