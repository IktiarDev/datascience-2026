```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import seaborn as sns

URL = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/penguins.csv"
df = pd.read_csv(URL).dropna()
print("Dataset shape:", df.shape)
```

```python
# -- 1. SETUP FIGURE LAYOUT --
fig = plt.figure(figsize=(16, 12))
fig.suptitle('Dashboard Analisis Biometrik Pinguin', fontsize=18, fontweight='bold', y=0.98)

# Membuat grid layout 2 x 2 untuk multi-subplot
gs = gridspec.GridSpec(2, 2, figure=fig, hspace=0.4, wspace=0.35)

ax1 = fig.add_subplot(gs[0, 0])  # Kiri atas (Bar Chart)
ax2 = fig.add_subplot(gs[0, 1])  # Kanan atas (Histogram)
ax3 = fig.add_subplot(gs[1, 0])  # Kiri bawah (Boxplot)
ax4 = fig.add_subplot(gs[1, 1])  # Kanan bawah (Scatter Plot)
```

```python
# -- 2. GRAFIK 1: Bar Chart (Countplot) --
sns.countplot(data=df, x='species', ax=ax1, palette='Set2', hue='species', legend=False)
ax1.set_title('Jumlah Pinguin per Spesies', fontsize=14, fontweight='bold')
ax1.set_xlabel('Spesies')
ax1.set_ylabel('Jumlah (Ekor)')
ax1.grid(True, axis='y', alpha=0.3)
```

### Insight Grafik 1 - Bar Chart
Spesies pinguin **Adelie** merupakan kelompok dengan jumlah ekor terbanyak dalam dataset (146 ekor), disusul oleh **Gentoo** (119 ekor), dan **Chinstrap** sebagai kelompok terkecil (68 ekor). Hal ini menunjukkan adanya ketidakseimbangan kelas (*class imbalance*) ringan dalam dataset populasi pinguin.

```python
# -- 3. GRAFIK 2: Histogram + KDE --
sns.histplot(data=df, x='flipper_length_mm', kde=True, color='teal', ax=ax2, bins=15)
ax2.set_title('Distribusi Panjang Sirip Pinguin', fontsize=14, fontweight='bold')
ax2.set_xlabel('Panjang Sirip (mm)')
ax2.set_ylabel('Frekuensi')
ax2.grid(True, alpha=0.3)
```

### Insight Grafik 2 - Histogram & KDE
Sebaran panjang sirip pinguin memperlihatkan pola sebaran bimodal yang mengindikasikan adanya pencampuran karakteristik fisik pinguin. Puncak pertama di sekitar 190 mm mewakili spesies Adelie dan Chinstrap, sedangkan puncak kedua di kisaran 215 mm mewakili spesies Gentoo yang berukuran lebih besar.

```python
# -- 4. GRAFIK 3: Boxplot --
sns.boxplot(data=df, x='island', y='bill_length_mm', ax=ax3, palette='Pastel1', hue='island', legend=False)
ax3.set_title('Sebaran Panjang Paruh Berdasarkan Pulau Hunian', fontsize=14, fontweight='bold')
ax3.set_xlabel('Pulau')
ax3.set_ylabel('Panjang Paruh (mm)')
ax3.grid(True, axis='y', alpha=0.3)
```

### Insight Grafik 3 - Boxplot
Pinguin di pulau **Torgersen** memiliki rata-rata panjang paruh terpendek dengan rentang variasi yang sempit. Sementara itu, pinguin di pulau **Biscoe** dan **Dream** memiliki rata-rata panjang paruh yang lebih panjang dan rentang variasi yang lebih lebar. Hal ini kemungkinan dipengaruhi oleh jenis spesies pinguin yang menghuni masing-masing pulau.

```python
# -- 5. GRAFIK 4: Scatter Plot --
sns.scatterplot(data=df, x='bill_length_mm', y='bill_depth_mm', hue='species', style='species', ax=ax4, s=60)
ax4.set_title('Hubungan Panjang vs Kedalaman Paruh Pinguin', fontsize=14, fontweight='bold')
ax4.set_xlabel('Panjang Paruh (mm)')
ax4.set_ylabel('Kedalaman Paruh (mm)')
ax4.grid(True, alpha=0.3)
```

### Insight Grafik 4 - Scatter Plot
Scatter plot memperlihatkan pengelompokan (klasterisasi) spesies yang sangat kentara. Pinguin **Gentoo** memiliki paruh panjang namun tipis/dangkal (*bill depth* rendah). Pinguin **Adelie** memiliki paruh pendek namun tebal/dalam. Sedangkan pinguin **Chinstrap** memiliki paruh panjang sekaligus tebal/dalam.

```python
# -- 6. EKSPOR DASHBOARD --
plt.savefig('dashboard_pinguin.png', bbox_inches='tight', dpi=150)
print("Dashboard berhasil disimpan ke file 'dashboard_pinguin.png'.")
plt.show()
```
